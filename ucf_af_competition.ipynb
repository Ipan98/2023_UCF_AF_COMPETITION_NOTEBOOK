However, we still have multiple entry for each of our users while only having one finale state prediction (Delinquant or Not Delinquant). Our problem seems to be a time series classification. We are going to need to need to keep the last entry for each features, in addition for our numerical features we are going to generate new features which are correlated to the actual features such as the standard deviation, the minimum and maximum encountered such that our model can capture the different status of that customer account

```python
features=[ 'CreditCard_id',  'CreditLine',
       'LastStatementBalanceAmount', 'LastStatementPurchaseAmount',
       'LastStatementPurchaseReturnAmount',
       'LastStatementMinimumPaymentDueAmount',
       'LastStatementPaymentTotalAmount',  'DelinquentTotalAmount',
       'LifetimeDelinquent1CycleCount', 'LifetimeDelinquent2CycleCount',
       'LifetimeDelinquent3CycleCount', 'LifetimeDelinquent4CycleCount',
       'LifetimeDelinquent5CycleCount', 'LifetimeDelinquent6CycleCount',
       'LifetimeDelinquent7CycleCount', 'ChargeOffAmount']
aggr=df[features].groupby("CreditCard_id").agg(['last', 'std', 'min', 'max'])
aggr.columns = ['_'.join(x) for x in aggr.columns]
aggr=aggr.reset_index()
aggr
```

For the categorical values, we will get the last entry and the mode also known as the most common values

```python
from sklearn.preprocessing import LabelEncoder
df['CardInternalStatus']= LabelEncoder().fit_transform(df['CardInternalStatus'])
df['ActivityHistory1_12']= LabelEncoder().fit_transform(df['ActivityHistory1_12'])
cat=['CreditCard_id', 'CardInternalStatus',
       'ActivityHistory1_12', 'Target']
aggr_cat=df[cat].groupby("CreditCard_id").agg(['last',lambda x: pd.Series.mode(x)[0]])#lambda x: pd.Series.mode(x)[0]
aggr_cat.columns = ['_'.join(x) for x in aggr_cat.columns]
aggr_cat=aggr_cat.reset_index()
aggr_cat=aggr_cat.drop(columns=['Target_<lambda_0>'])
aggr_cat.columns=['CreditCard_id', 'CardInternalStatus_last',
       'CardInternalStatus_mode', 'ActivityHistory1_12_last',
       'ActivityHistory1_12_mode', 'Target_last']
aggr_cat
```

One thing we keep noticing is the change in the last 3 months before the last delinquancy entry, we should create features that should be able to capture that. We can do that by using the Exponential Moving Average(EMA) with the last 3 months entries. We are using the EMA as it applies more weight to later data

```python
#SSN_id CreditCard_id
features=['CreditCard_id',  'CreditLine',
       'LastStatementBalanceAmount', 'LastStatementPurchaseAmount',
       'LastStatementPurchaseReturnAmount',
       'LastStatementMinimumPaymentDueAmount',
       'LastStatementPaymentTotalAmount', 'DelinquentTotalAmount',
       'LifetimeDelinquent1CycleCount', 'LifetimeDelinquent2CycleCount',
       'LifetimeDelinquent3CycleCount', 'LifetimeDelinquent4CycleCount',
       'LifetimeDelinquent5CycleCount', 'LifetimeDelinquent6CycleCount',
       'LifetimeDelinquent7CycleCount', 'ChargeOffAmount']
df_ema=df[features].groupby("CreditCard_id").ewm(span=3).mean()#24
df_ema=df_ema.reset_index()
df_ema=df_ema.drop(columns=['level_1'])
df_ema=df_ema.groupby("CreditCard_id").agg(['last'])
df_ema.columns = ['_ema_'.join(x) for x in df_ema.columns]
df_ema=df_ema.reset_index()
df_ema
```

After creating all those features, we can finally create our final dataFrame

```python
df_merge = pd.merge(aggr, df_ema, how='outer', on=['CreditCard_id'])
df_merge = pd.merge(df_merge, aggr_cat, how='outer', on=['CreditCard_id'])
df_merge
```

We end up with 81 columns, and around 79 features, let's build a model. We are going to use Light GBM with a 0.03, using Dart tree, 10000 estimators, then 15000 estimators

```python
from sklearn.metrics import roc_auc_score
from sklearn.metrics import accuracy_score
search_params = { 
    'learning_rate' : 0.03, #0.065,
    #'lambda_l1': 8.481514197781607,
    'lambda_l2': 50, #0.0004266339834880936,
    'num_leaves': 100, #27,
    'feature_fraction': 0.19, #0.484, 0.3
    #'bagging_fraction': 0.8477112190030014,
    #'bagging_freq': 2,
    'min_child_samples': 2400 #40 #20
}

fixed_params={
    'objective': 'binary',
    'metric': 'custom', 
    'boosting_type' : 'dart',
    #'force_row_wise' : True,
    #'device': 'gpu',
    #'n_jobs': -1,
    #'extra_trees' : True,
    #'feature_pre_filter': False,
    'n_estimators': 15000, 
    'early_stopping_round': 100
}
import lightgbm as lgb
import numpy as np
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score,roc_auc_score
from lightgbm import LGBMClassifier, early_stopping, log_evaluation
y=df_merge['Target_last']
X=df_merge.iloc[:, 1:-1]
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.25, random_state=42)
model = LGBMClassifier(**fixed_params, **search_params)
model.fit(X_train,y_train,eval_set=[(X_test,y_test),(X_train,y_train)],
          verbose=50,eval_metric='logloss', callbacks=[early_stopping(50), log_evaluation(0)])
predict=model.predict_proba(X_test)
predict = np.transpose([pred[1] for pred in predict])
print(roc_auc_score(y_test, predict))
y_pred=model.predict(X_test)
print(accuracy_score(y_test, y_pred))
```

```python
from sklearn.preprocessing import LabelEncoder
def features(df):
    df=df.sort_values(['CreditCard_id','MonthEndDate' ],ascending=True)
    features=[ 'CreditCard_id',  'CreditLine',
       'LastStatementBalanceAmount', 'LastStatementPurchaseAmount',
       'LastStatementPurchaseReturnAmount',
       'LastStatementMinimumPaymentDueAmount',
       'LastStatementPaymentTotalAmount',  'DelinquentTotalAmount',
       'LifetimeDelinquent1CycleCount', 'LifetimeDelinquent2CycleCount',
       'LifetimeDelinquent3CycleCount', 'LifetimeDelinquent4CycleCount',
       'LifetimeDelinquent5CycleCount', 'LifetimeDelinquent6CycleCount',
       'LifetimeDelinquent7CycleCount', 'ChargeOffAmount']
    aggr=df[features].groupby("CreditCard_id").agg(['last', 'std', 'min', 'max'])
    aggr.columns = ['_'.join(x) for x in aggr.columns]
    aggr=aggr.reset_index()
    from sklearn.preprocessing import LabelEncoder
    df['CardInternalStatus']= LabelEncoder().fit_transform(df['CardInternalStatus'])
    df['ActivityHistory1_12']= LabelEncoder().fit_transform(df['ActivityHistory1_12'])
    cat=['CreditCard_id', 'CardInternalStatus',
       'ActivityHistory1_12', 'Target']
    aggr_cat=df[cat].groupby("CreditCard_id").agg(['last',lambda x: pd.Series.mode(x)[0]])#lambda x: pd.Series.mode(x)[0]
    aggr_cat.columns = ['_'.join(x) for x in aggr_cat.columns]
    aggr_cat=aggr_cat.reset_index()
    aggr_cat=aggr_cat.drop(columns=['Target_<lambda_0>'])
    aggr_cat.columns=['CreditCard_id', 'CardInternalStatus_last',
       'CardInternalStatus_mode', 'ActivityHistory1_12_last',
       'ActivityHistory1_12_mode', 'Target_last']
    #SSN_id CreditCard_id
    features=['CreditCard_id',  'CreditLine',
       'LastStatementBalanceAmount', 'LastStatementPurchaseAmount',
       'LastStatementPurchaseReturnAmount',
       'LastStatementMinimumPaymentDueAmount',
       'LastStatementPaymentTotalAmount', 'DelinquentTotalAmount',
       'LifetimeDelinquent1CycleCount', 'LifetimeDelinquent2CycleCount',
       'LifetimeDelinquent3CycleCount', 'LifetimeDelinquent4CycleCount',
       'LifetimeDelinquent5CycleCount', 'LifetimeDelinquent6CycleCount',
       'LifetimeDelinquent7CycleCount', 'ChargeOffAmount']
    df_ema=df[features].groupby("CreditCard_id").ewm(span=3).mean()#24
    df_ema=df_ema.reset_index()
    df_ema=df_ema.drop(columns=['level_1'])
    df_ema=df_ema.groupby("CreditCard_id").agg(['last'])
    df_ema.columns = ['_ema_'.join(x) for x in df_ema.columns]
    df_ema=df_ema.reset_index()
    df_merge = pd.merge(aggr, df_ema, how='outer', on=['CreditCard_id'])
    df_merge = pd.merge(df_merge, aggr_cat, how='outer', on=['CreditCard_id'])
    return df_merge
```
